```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
from statsmodels.stats.outliers_influence import variance_inflation_factor
import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully.")
```

```python
file_path = "diabetes_data.xls"

df = pd.read_csv(file_path, sep=r"\s+", header=None)

print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
feature_names = [
    "age", "sex", "bmi", "bp",
    "s1", "s2", "s3", "s4", "s5", "s6"
]

df.columns = feature_names

print("Column names:")
print(df.columns.tolist())
```

```python
diabetes = load_diabetes()

y = pd.Series(diabetes.target, name="target")

print("Target Shape:", y.shape)
print("\nFirst 10 target values:")
print(y.head(10).to_list())
```

```python
X = df.copy()

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nMissing values:")
print(X.isnull().sum().sum())

print("\nTarget missing values:")
print(y.isnull().sum())
```

```python
vif_data = pd.DataFrame()

vif_data["Feature"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(X.values.astype(float), i)
    for i in range(X.shape[1])
]

print("Variance Inflation Factor (VIF):")
display(vif_data)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully.")
```

```python
ols = LinearRegression()

ols.fit(X_train_scaled, y_train)

y_pred_ols = ols.predict(X_test_scaled)

ols_mse = mean_squared_error(y_test, y_pred_ols)
ols_r2 = r2_score(y_test, y_pred_ols)

print("OLS Regression")
print("-------------------------")
print("MSE :", round(ols_mse, 4))
print("R2  :", round(ols_r2, 4))

print("\nOLS Coefficients:")
for feature, coef in zip(feature_names, ols.coef_):
    print(f"{feature:5s}: {coef:.4f}")
```

```python
ridge = Ridge(alpha=1.0)

ridge.fit(X_train_scaled, y_train)

y_pred_ridge = ridge.predict(X_test_scaled)

ridge_mse = mean_squared_error(y_test, y_pred_ridge)
ridge_r2 = r2_score(y_test, y_pred_ridge)

print("Ridge Regression")
print("-------------------------")
print("Alpha:", 1.0)
print("MSE  :", round(ridge_mse, 4))
print("R2   :", round(ridge_r2, 4))

print("\nRidge Coefficients:")
for feature, coef in zip(feature_names, ridge.coef_):
    print(f"{feature:5s}: {coef:.4f}")
```

```python
ridge = Ridge(alpha=1.0)

ridge.fit(X_train_scaled, y_train)

y_pred_ridge = ridge.predict(X_test_scaled)

ridge_mse = mean_squared_error(y_test, y_pred_ridge)
ridge_r2 = r2_score(y_test, y_pred_ridge)

print("Ridge Regression")
print("-------------------------")
print("Alpha:", 1.0)
print("MSE  :", round(ridge_mse, 4))
print("R2   :", round(ridge_r2, 4))

print("\nRidge Coefficients:")
for feature, coef in zip(feature_names, ridge.coef_):
    print(f"{feature:5s}: {coef:.4f}")
```

```python
lasso = Lasso(alpha=1.0, max_iter=100000)

lasso.fit(X_train_scaled, y_train)

y_pred_lasso = lasso.predict(X_test_scaled)

lasso_mse = mean_squared_error(y_test, y_pred_lasso)
lasso_r2 = r2_score(y_test, y_pred_lasso)

print("Lasso Regression")
print("-------------------------")
print("Alpha:", 1.0)
print("MSE  :", round(lasso_mse, 4))
print("R2   :", round(lasso_r2, 4))

print("\nLasso Coefficients:")
for feature, coef in zip(feature_names, lasso.coef_):
    print(f"{feature:5s}: {coef:.4f}")

print("\nNumber of non-zero coefficients:",
      np.count_nonzero(lasso.coef_))
```

```python
alphas = np.logspace(-2, 2, 50)

ridge_results = []
lasso_results = []

for alpha in alphas:

    # Ridge
    ridge_model = Ridge(alpha=alpha)
    ridge_model.fit(X_train_scaled, y_train)

    ridge_pred = ridge_model.predict(X_test_scaled)

    ridge_results.append({
        "Alpha": alpha,
        "MSE": mean_squared_error(y_test, ridge_pred),
        "R2": r2_score(y_test, ridge_pred),
        "Non-zero": np.count_nonzero(ridge_model.coef_)
    })

    # Lasso
    lasso_model = Lasso(alpha=alpha, max_iter=100000)
    lasso_model.fit(X_train_scaled, y_train)

    lasso_pred = lasso_model.predict(X_test_scaled)

    lasso_results.append({
        "Alpha": alpha,
        "MSE": mean_squared_error(y_test, lasso_pred),
        "R2": r2_score(y_test, lasso_pred),
        "Non-zero": np.count_nonzero(lasso_model.coef_)
    })

ridge_results = pd.DataFrame(ridge_results)
lasso_results = pd.DataFrame(lasso_results)

print("Ridge Results:")
display(ridge_results.head())

print("Lasso Results:")
display(lasso_results.head())
```

```python
best_ridge = ridge_results.loc[ridge_results["R2"].idxmax()]
best_lasso = lasso_results.loc[lasso_results["R2"].idxmax()]

print("Best Ridge Model")
print("-------------------------")
print("Alpha     :", round(best_ridge["Alpha"], 4))
print("MSE       :", round(best_ridge["MSE"], 4))
print("R2        :", round(best_ridge["R2"], 4))
print("Non-zero  :", int(best_ridge["Non-zero"]))

print("\nBest Lasso Model")
print("-------------------------")
print("Alpha     :", round(best_lasso["Alpha"], 4))
print("MSE       :", round(best_lasso["MSE"], 4))
print("R2        :", round(best_lasso["R2"], 4))
print("Non-zero  :", int(best_lasso["Non-zero"]))
```

```python
comparison = pd.DataFrame({
    "Model": ["OLS", "Ridge", "Lasso"],
    "MSE": [ols_mse, ridge_mse, lasso_mse],
    "R2": [ols_r2, ridge_r2, lasso_r2],
    "Non-zero Coefficients": [
        np.count_nonzero(ols.coef_),
        np.count_nonzero(ridge.coef_),
        np.count_nonzero(lasso.coef_)
    ]
})

comparison["MSE"] = comparison["MSE"].round(4)
comparison["R2"] = comparison["R2"].round(4)

display(comparison)
```
